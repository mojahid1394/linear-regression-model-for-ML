```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
```

```python
# load the diabetes dataset
daibetes = load_diabetes()
```

```python
df = pd.DataFrame(daibetes.data, columns=daibetes.feature_names)
df
```

```python
df['target'] = daibetes.target
df
```

```python
df.head()
```

```python
X  = df.drop('target', axis = 1)
X
```

```python
y = df['target']
y
```

```python
X.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg
```

```python
# train the model on the training set - this is how the model learn
reg.fit(X_train, y_train)
```

```python
y_pred = reg.predict(X_test)
y_pred
```

```python
y_pred.shape
```

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

```python
r2 = r2_score(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
intercept = reg.intercept_
```

```python
print('R2: ', r2)
print('mse: ', mse)
print('mae: ', mae)
print('intercept: ', intercept)
```

```python
reg.score(X_test, y_test)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(y_test, y_pred, alpha=0.5)
plt.plot([y.min(), y.max()], [y.min(), y.max()], color='red')
plt.xlabel('Actual values')
plt.ylabel('Predicted values')
plt.title('Predicted vs Actual values')
plt.show()
```

```python
# residuals(loss) = y
residual = y_test - y_pred
residual
```

```python
plt.scatter(y_pred, residual, alpha=0.5)
plt.xlabel('Predicted values')
plt.ylabel('Residual')
plt.title('Residual Plot')
plt.show()
```
